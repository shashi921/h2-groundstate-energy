```python
input = """2
Sample H2 molecule
H 0.3710 0.0 0.0
H -0.3710 0.0 0.0"""
```

```python
import xyz_parse
import pennylane as qml
from pennylane import numpy as np

molecule = xyz_parse.Molecule.parse(input)
H, _ = qml.qchem.molecular_hamiltonian(
    molecule.symbols,
    np.array(molecule.coordinates, dtype=np.float64) * 1.88973
)

# Direct Hermitian eigenvalue calculation (minimal overhead)
evals = np.linalg.eigvalsh(qml.matrix(H))
output = float(evals[0])
```
